```python
from sqlalchemy import create_engine
import pandas as pd, getpass

engine = create_engine(f'postgresql://{getpass.getpass("User for the server: ")}:{getpass.getpass("Passwrod for postgres server:")}@localhost:5432/{input("DB name:")}')

try:
    connection = engine.connect()
    print("Connection to PostgreSQL established successfully!")
except Exception as e:
    print(f"Connection failed! Error: {e}")
```

First, we retrieve the types of data we have

```python
from sqlalchemy import text

with engine.connect() as conn:
    result = conn.execute(text('''
        select table_name
        from information_schema.tables
        where table_schema='public'
        ;
    '''))
    list_tables = [row[0] for row in result.fetchall()]
```

```python
print(f"{'TABLE':<23}| COLUMN(TYPE)")
print('-'*150)
for name in list_tables:
    query = text(f'''
        select column_name, data_type 
        from information_schema.columns
        where table_name='{name}';
    ''')

    with engine.connect() as conn:
        result = conn.execute(query)

        columns = result.mappings().all()
        
        print(f'{name:<23}| '+', '.join([f'{col["column_name"]}({col["data_type"].upper()})' for col in columns]))
```

```python
query = '''
        select id,country
        from public.countries;
'''

country_df = pd.read_sql(query, engine)
country_df
```

```python
query = '''
    with cases as (
        select date,new_cases,new_deaths,country
        from public.covid_cases as cases
        join public.countries as countries
        on cases.country_id=countries.id
        where countries.country='italy')
    ,
    italy_weekly_cases as (
    select date,(date-first_date.day)/7 +1 as week, new_cases
    FROM italy_cases
    cross join (select min(date) as first_day, country_id
        from italy_cases) as first_date
    )

    select max(date) as date, week, sum(new_cases) as new_cases
    from italy_weekly_cases
    group by week
    order by week;
'''

new_cases_italy = pd.read_sql(query, engine)

if 'date' in new_cases_italy.columns:
    new_cases_italy['date'] = new_cases_italy['date'].astype('datetime64[us]')

new_cases_italy
```
