# Analysis of Covid data

**My goal**: To use more of SQL/PostgreSQL instead of Pandas, in order to improve my skills.

I first load the data on a weekly basis until the 5th of May 2023 ([WHO - end of pandemic](https://www.sciencedirect.com/science/article/pii/S1876034125000474#bib2)), then **scale all the reported numbers** (e.g number of new weekly cases) by the population of a country in millions ($\equiv$ value / population_in_millions). This way, countries of vastly different sizes such as Switzerland and United States can be compared more reliably.


My goal is to answer the following questions/perform analysis such as:

1. What are the overall trends in the number of new cases and deaths in a country throughout the pandemic? Especially regarding COVID variants.

2. A comparison between CFR and fatality rates of patients after 2, 4, and 8 weeks.

         The variable CFR is defined as (Confirmed Deaths up to nth week / Confirmed Cases up to nth week). The reason for the time lags is to account for the time between onset of the disease and reported death available on the [CDC website](https://www.cdc.gov/nchs/nvss/vsrr/covid19/index.htm). However, even with a time lag, these rates still suffer from other shortcomings attributed to CFR, such as, not accounting for limited testing and the varying risk between different demographic groups and patients' condition of living [doi:10.1016/S1473-3099(10)70120-1](https://doi.org/10.1016/S1473-3099(10)70120-1).

3. Does the administration of vaccines affect these numbers? Both in terms of raw numbers and the correlation between vaccine numbers and new case/deaths.
 
***Note:*** We do not investigate the lockdowns effects since in each countries, the lockdowns were erratically introduced and halted. Therefore, it is difficult to show case them on a 2D plot. Potentially, this data could be included in the interactive dashboards (*future plans*).

## 1. What are the overall trends in the number of new cases and deaths in a country throughout the pandemic?

Explanation of the queries:

- first_monday --> auxilliary for calculating number of weeks for the subsequent queries
  
- weekly --> new cases and deaths reported, evaluated on a weekly basis, and cumulative sum of them, to determine all_cases and deaths

- Retrieve the weekly, new_cases, deaths, all cases, all deaths.

```python
qry = '''
        with first_monday as ( 
        select 
            min(date) as monday
        from public.covid_cases_deaths
        where extract(dow from date)=1        
    )
    , 
    weekly as (
             select 
                min(cases.date) as firstday_per_week,
                weeks.week as week,
                cases.country_id as country_id,
                sum(cases.new_cases) as new_cases,
                sum(cases.new_deaths) as new_deaths,
                max(weeks.all_cases) as all_cases,
                max(weeks.all_deaths) as all_deaths
                
        from public.covid_cases_deaths as cases
        left join (select 
                cases.date,
                (cases.date-first_monday.monday)/7 +1 as week,
                sum(cases.new_cases) over (partition by cases.country_id
                                            order by cases.date) as all_cases,
                sum(cases.new_deaths) over (partition by cases.country_id
                                            order by cases.date) as all_deaths,
                country_id
                
                from public.covid_cases_deaths as cases
                cross join first_monday
                order by cases.country_id, cases.date
            ) as weeks
        on cases.country_id=weeks.country_id and cases.date=weeks.date
        group by cases.country_id, weeks.week
        order by cases.country_id, weeks.week
    )
    

    select 
            firstday_per_week,
            week,
            country_id,
            new_cases,
            new_deaths,
            all_cases,
            all_deaths            
    from weekly
    where weekly.firstday_per_week <= '2023-05-05'
    order by weekly.country_id, weekly.week
    ;
        '''
with engine.begin() as conn:
    cases_deaths_df = pd.read_sql(qry, conn)
cases_deaths_df
```

```python
xticks = []
xticks_labels = []

def season(day):
    if day in pd.date_range(f"{day.year}-04-01",f"{day.year}-06-30"):
        return "Spring"
    elif day in pd.date_range(f"{day.year}-07-01",f"{day.year}-09-30"):
        return "Summer"
    elif day in pd.date_range(f"{day.year}-10-01",f"{day.year}-12-31"):
        return "Autumn"
    else:
        return "Winter"

time_df = cases_deaths_df[["firstday_per_week","week"]]
time_df["year"] = time_df["firstday_per_week"].transform(lambda x: x.year)
time_df["season"] = time_df["firstday_per_week"].astype("datetime64[us]").transform(season)

time_df = time_df.groupby(["season","year"])["week"].min().reset_index().sort_values(by="week")
xticks = time_df.week.values
xticks_labels = (time_df.year.astype("str")+" ("+time_df.season+")").values

########################################
import matplotlib as mpl

n_colors = len(country_df.country.unique())
cmap = mpl.colormaps['tab20']

colors = [mpl.colors.rgb2hex(i) for i in cmap(np.linspace(0, 1, n_colors))]
color_map = dict(zip(country_df["id"], colors))
###############################################

linewidths = [3 for i in range(5)]+[2.5 for i in range(5)]+[2 for i in range(5)]+[1.5 for i in range(5)]
```

## 2. A comparison between CFR and the fatality rates of patients after 2, 4, and 8 weeks.

- lagged_cases --> Pushing the cases columns 2, 4, and 8 rows further. Zero values are set to Null (not to encounter 0 in the denominator in our ratio calc.)
- r_?? --> ratio of all
- and the ratio between the last two, after 2, 4, and 8 weeks. The idea is to estimate the average estimate of mortality time of covid, following these articles: [paper1](), [paper2](), [paper3]() . We expect a plateau in the ratio would be a good indicator of such quantity.

```python
qry = '''
        with first_monday as ( 
        select 
            min(date) as monday
        from public.covid_cases_deaths
        where extract(dow from date)=1        
    )
    , 
    weekly as (
             select 
                min(cases.date) as firstday,
                weeks.week as week,
                cases.country_id as country_id,
                sum(cases.new_cases) as new_cases,
                sum(cases.new_deaths) as new_deaths,
                max(weeks.all_cases) as all_cases,
                max(weeks.all_deaths) as all_deaths
                
        from public.covid_cases_deaths as cases
        left join (select 
                cases.date,
                (cases.date-first_monday.monday)/7 +1 as week,
                sum(cases.new_cases) over (partition by cases.country_id
                                            order by cases.date) as all_cases,
                sum(cases.new_deaths) over (partition by cases.country_id
                                            order by cases.date) as all_deaths,
                country_id
                
                from public.covid_cases_deaths as cases
                cross join first_monday
                order by cases.country_id, cases.date
            ) as weeks
        on cases.country_id=weeks.country_id and cases.date=weeks.date
        group by cases.country_id, weeks.week
        order by cases.country_id, weeks.week
    ),
    lagged_cases as (
        select firstday,
                week,
                country_id,
                /*
                        We need the nullif, since otherwise we'd have 0 denominators in our ratio calc. below
                */
                nullif(
                lag(all_cases, 2) over (
                                        partition by country_id
                                        order by week
                                        ),0) as lagged_2weeks,
                
                nullif(
                lag(all_cases, 4) over (
                                        partition by country_id
                                        order by week
                                        ),0) as lagged_4weeks,
                
                nullif(
                lag(all_cases, 8) over (
                                        partition by country_id
                                        order by week
                                        ),0) as lagged_8weeks
        from weekly
        order by country_id, week
    )
    

    select 
            weekly.firstday as firstday_per_week,
            weekly.week as week,
            weekly.country_id as country_id,
            weekly.all_cases as all_cases,
            weekly.all_deaths as all_deaths,
            coalesce(
                    weekly.all_deaths/nullif(weekly.all_cases,0), 0)
                        as r_cfr,
            coalesce(
                    weekly.all_deaths/lagged_cases.lagged_2weeks, 0)
                        as r_lag_2w, 	
            coalesce(
                    weekly.all_deaths/lagged_cases.lagged_4weeks, 0)
                        as r_lag_4w, 	
            coalesce(
                    weekly.all_deaths/lagged_cases.lagged_8weeks, 0)
                        as r_lag_8w
            
    from weekly
    join lagged_cases
    on weekly.firstday=lagged_cases.firstday and 
        weekly.week=lagged_cases.week and 
        weekly.country_id=lagged_cases.country_id
    where weekly.firstday <= '2023-05-05'
    order by weekly.country_id, weekly.week
    ;
        '''

mortality_rates_df = pd.read_sql(qry, engine)
mortality_rates_df
```

```python
fig = plt.figure(figsize=[15,15])
gs = plt.GridSpec(4, 2, width_ratios=[1.,0.05], height_ratios=[1,1,1,1] , figure=fig, hspace=0.05)

ax = [fig.add_subplot(gs[i,0], sharex=None if i==0 else fig.axes[0]) for i in range(gs.get_geometry()[0])]
for (idx,name,population) in country_df[["id","country","population_mil"]].values:
    data = mortality_rates_df.loc[mortality_rates_df.country_id==idx]
    for counter,y_ in enumerate(["r_cfr","r_lag_2w","r_lag_4w","r_lag_8w"]):
        data.loc[data[y_]==0,y_] = np.nan
        sns.lineplot(x="week",
                     y=y_,
                     data=data,
                     ax = ax[counter],
                     color=color_map[idx],
                     linewidth=linewidths[idx-1],
                     label = name+f" (#{idx})")

labels = ["(CFR)", "(2 weeks Lag)", "(4 weeks Lag)", "(8 weeks Lag)"]
for counter in range(gs.get_geometry()[0]):
    ax[counter].get_legend().remove()
    ax[counter].set_ylim(0,0.1)
    ax[counter].set_yticks([i/100 for i in range(1,10)])
    ax[counter].set_xticks(xticks)
    ax[counter].set_xlabel("")
    ax[counter].tick_params(axis="x", labelbottom=False)
    ax[counter].set_ylabel('Moratlity rate '+labels[counter])

ax[-1].set_xticks(xticks,xticks_labels, rotation=80)
ax[-1].tick_params(axis="x", labelbottom=True)

ax_country_list = fig.add_subplot(gs[:,1])
handles = [
    plt.Line2D([0], [0], color=line.get_color(), lw=2, label=' '.join([i.capitalize() for i in line.get_label().split()]))
    for idx,line in enumerate(ax[0].get_lines())
]

ax_country_list.legend(
    handles=handles,
    title="Country (rank GDP per cap.)",
    loc="center left",
    frameon=False,
    fontsize=10,
    handlelength=1,
    labelspacing=1.0
)

# Hide the legend axis
ax_country_list.axis("off")
```
